# 03 - Inference
Load the saved pipeline and score patients locally, then (optionally) call the running API.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))
import os
os.chdir(ROOT)
```

```python
import json
from heart.predict import load_model, predict_records
model = load_model()
model
```

```python
patients = [json.load(open(p)) for p in ['samples/sample_request.json', 'samples/sample_request_low_risk.json']]
predict_records(model, patients)
```

Missing `ca` / `thal` are allowed - the pipeline imputes them exactly as in training.

```python
predict_records(model, [{**patients[0], 'ca': None, 'thal': None}])
```

## Call the API (start it first with `make serve` or the container)

```python
import urllib.request
try:
    req = urllib.request.Request('http://127.0.0.1:8000/predict', json.dumps(patients[0]).encode(), {'Content-Type': 'application/json'})
    print(json.loads(urllib.request.urlopen(req, timeout=3).read()))
except OSError as err:
    print('API not running:', err)
```
